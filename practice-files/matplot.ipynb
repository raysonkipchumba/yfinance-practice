```python
import matplotlib.pyplot as plt
```

```python
#Scatter Plot

x = [1, 2, 3, 4, 5]
y = [2, 4, 5, 4, 8]
plt.scatter(x, y, color='purple')
plt.title('Scatter Plot Example')
plt.show()
```

```python
# HISTOGRAM
data = [12, 15, 12, 18, 22, 25, 30, 22, 19, 15]
plt.hist(data, bins=5, color='green', edgecolor='black')
plt.title('Histogram Example')
plt.show()
```

```python
#PIE CHART
sizes = [40, 30, 20, 10]
labels = ['Category A', 'Category B', 'Category C', 'Category D']
plt.pie(sizes, labels=labels, autopct='%1.1f%%')
plt.title('Pie Chart Example')
plt.show()
```

```python
# BOX CHART
data = [10, 15, 18, 22, 25, 28, 40] # 40 might show up as an outlier
plt.boxplot(data)
plt.title('Box Plot Example')
plt.show()
```

```python
# Bar Graph
plt.bar(['Apples', 'Oranges'], [15, 30], color=['skyblue', 'salmon'])
plt.title('Bar Graph')
plt.show()
```

```python
# Line Graph
plt.plot(['Day 1', 'Day 2', 'Day 3', 'Day 4'], [10, 25, 18, 32], marker='o', color='b')
plt.title('Line Graph')
plt.show()
```
